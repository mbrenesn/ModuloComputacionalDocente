# Paralelismo de memoria distribuida

El paralelismo de memoria distribuida es un paradigma de concurrencia que aplica cuando tenemos unidades lógicas de procesamiento que **completamente independientes** con **bancos de memoria independientes**, i.e., se pueden sincronizar elementos de computación mediante un protocolo de **pasos de mensajes**.

    * Cada unidad se divide en una serie de instrucciones.

## ¿Cómo evitar un deadlock?

* La idea corresponde en **serializar** el procedimiento que no puede realizarse de forma concurrente en una **operación de cascada**.

<div style="text-align: center;">
<img src="Figs/Fig6.png" width="420">

* El problema principal de serializar procedimientos corresponde a que **añaden al tiempo serial** que domina la eficiencia del programa (recordar ley de Amhdal).
* Por esta razón, a la medida de lo posible se debe:
    * reducir eventos de comunicación y
    * agrupar comunicaciones pequeñas en un solo mensaje grande.
* Adicionalmente, se debe eliminar sincronización siempre que sea posible dado que reduce el desempeño a aquel del proceso mas lento.

## Diseño de programa: MPI (*Message Passing Interface*)

* Procesos múltiples y separados (los cuales pueden ser locales o remotos) son coordinados de manera concurrente para intercambiar datos a través de **mensajes**.
    * Los procesos, en principio, no comparten memoria con otros procesos.
* La idea es distribuir campos de datos muy grandes, replicando la menor cantidad de información.

#### MPI

* MPI es un estándar:
    * Existe documentación para describir como las funciones de la API deben ser llamadas y su comportamiento general.
    * Existen tres niveles: MPI-1 (básico), MPI-2 (intermedio) y MPI-3 (avanzado)
    * [Documentación general](http://www.mpi-forum.org)
    * [Tutoriales de funcionalidad básica](https://mpitutorial.com/)
* Las implementaciones de MPI se utilizan para esconder los detalles de la comunicación que ocurre a nivel de hardware.
* Existen diversas implementaciones, e.g., OpenMPI, MPICH e Intel MPI.

## Comunicadores

* Los comunicadores son el objeto fundamental que provee MPI.
    * La comunicación y el paso de mensajes ocurre a través de un comunicador.
    * Las fuentes y los destinatarios se identifican con un **rango**.

<div style="text-align: center;">
<img src="Figs/Fig7.png" width="420">

* Por defecto, el comunicador principal involucra a todos los procesos que son invocados al momento de la ejecución.
    * `size`: el número de procesos MPI.
    * `rank`: el ID del proceso dentro del grupo de comunicación.

#### Hola, mundo!

* Ver `hello.cpp`
* Compilación: `mpicxx hello.cpp -o hello.x`
* Ejecución: `mpirun -np X ./hello.x` donde `X` es el número de procesos a utilizar.
